```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import Dense
from sklearn.model_selection import KFold

import pandas as pd
import numpy as np
```

```python
df = pd.read_csv('/content/data/sonar3.csv')
df.info()
```

```python
X = df.iloc[:, 0:60]
y = df.iloc[:, 60]
```

```python
k = 5

kfold = KFold(n_splits=k, shuffle=True)
acc_score_basic = []
acc_score = []
```

```python
def model_fn_basic():
  model = Sequential()
  model.add(Dense(24, input_dim=60, activation='relu'))
  model.add(Dense(10, activation='relu'))
  model.add(Dense(1, activation='sigmoid'))
  model.summary()
  return model
```

[Step 1] 베이스라인 모델 실행

```python
for train_idx, test_idx in kfold.split(X):
  X_train, X_test = X.iloc[train_idx, :], X.iloc[test_idx, :]
  y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

  model = model_fn_basic()
  model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
  model.fit(X_train, y_train, epochs=200, batch_size=10)
  accuracy = model.evaluate(X_test, y_test)[1]
  acc_score_basic.append(accuracy)
```

```python
acc_score_basic
```

```python
sum(acc_score_basic) / k
```

[Step 2] 모델 구조 변경 실험

```python
# 노드 수 변경 24, 10, 1 -> 64, 32, 1
def model_fn_1():
  model = Sequential()
  model.add(Dense(64, input_dim=60, activation='relu'))
  model.add(Dense(32, activation='relu'))
  model.add(Dense(1, activation='sigmoid'))
  model.summary()
  return model
```

```python
# 노드 수 변경 24, 10, 1 -> 16, 8, 1
def model_fn_2():
  model = Sequential()
  model.add(Dense(16, input_dim=60, activation='relu'))
  model.add(Dense(8, activation='relu'))
  model.add(Dense(1, activation='sigmoid'))
  model.summary()
  return model
```

```python
# 은닉층 수 변경 2개 -> 4개
def model_fn_3():
  model = Sequential()
  model.add(Dense(24, input_dim=60, activation='relu'))
  model.add(Dense(16, activation='relu'))
  model.add(Dense(10, activation='relu'))
  model.add(Dense(6, activation='relu'))
  model.add(Dense(1, activation='sigmoid'))
  model.summary()
  return model
```

```python
# 은닉층 수 변경 2개 -> 1개
def model_fn_4():
  model = Sequential()
  model.add(Dense(24, input_dim=60, activation='relu'))
  model.add(Dense(1, activation='sigmoid'))
  model.summary()
  return model
```

```python
# 활성화 함수 변경 Relu -> tanh
def model_fn_5():
  model = Sequential()
  model.add(Dense(24, input_dim=60, activation='tanh'))
  model.add(Dense(10, activation='tanh'))
  model.add(Dense(1, activation='sigmoid'))
  model.summary()
  return model
```

```python
model_functions = [model_fn_1, model_fn_2, model_fn_3, model_fn_4, model_fn_5]
```

```python
for model_fn in model_functions:
  for train_idx, test_idx in kfold.split(X):
    X_train, X_test = X.iloc[train_idx, :], X.iloc[test_idx, :]
    y_train, y_test = y.iloc[train_idx], y.iloc[test_idx]

    model = model_fn()
    model.compile(loss='binary_crossentropy', optimizer='adam', metrics=['accuracy'])
    model.fit(X_train, y_train, epochs=200, batch_size=10)
    accuracy = model.evaluate(X_test, y_test)[1]
    acc_score.append(accuracy)
```

[Step 3] 결과 비교표 작성

```python
acc_score_1, acc_score_2, acc_score_3, acc_score_4, acc_score_5 = [acc_score[i:i+5] for i in range(0, 25, 5)]

all_scores = [acc_score_1, acc_score_2, acc_score_3, acc_score_4, acc_score_5]

# 노드 수 변경 24, 10, 1 -> 64, 32, 1
# 노드 수 변경 24, 10, 1 -> 16, 8, 1
# 은닉층 수 변경 2개 -> 4개
# 은닉층 수 변경 2개 -> 1개
# 활성화 함수 변경 Relu -> tanh
cell = {
    '모델 번호': [1, 2, 3, 4, 5],
    '모델 구조': ["64-32-1", "16-8-1", "24-16-10-6-1", "24-1", "24-10-1"],
    '활성화 함수': ["relu-relu-sigmoid", "relu-relu-sigmoid", "relu-relu-sigmoid", "relu-relu-sigmoid", "tanh-tanh-sigmoid"],
    # 인덱스를 [0]~[4]로 수정 (파이썬은 0부터 시작하니까요!)
    'fold1': [s[0] for s in all_scores],
    'fold2': [s[1] for s in all_scores],
    'fold3': [s[2] for s in all_scores],
    'fold4': [s[3] for s in all_scores],
    'fold5': [s[4] for s in all_scores],
    '평균 정확도': [np.mean(s) for s in all_scores],
    '표준편차': [np.std(s) for s in all_scores]
}

# 데이터프레임으로 변환하여 표 출력
df_cell = pd.DataFrame(cell)
print(df_cell)
```
